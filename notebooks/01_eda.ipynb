# SkyCast — 01 Exploratory Data Analysis

This notebook analyzes the Indian airfare dataset distributions, pricing structures across airlines and cabin classes, and investigates why cabin class dominates feature importance.


```python
import pandas as pd
import numpy as np

# Load datasets
clean_df = pd.read_csv('Clean_Dataset.csv')
new_df = pd.read_csv('Flight-Price-Prediction-using-Machine-Learning-Random-Forest-XGBoost--main/IndianFlightdata - Sheet1.csv')
print(f'Clean dataset shape: {clean_df.shape}')
print(f'New Indian flight dataset shape: {new_df.shape}')
```

## 1. Cabin Class Pricing Disparity
Investigate why `class` accounts for ~89% of tree variance reduction.


```python
class_stats = clean_df.groupby('class')['price'].describe()
print(class_stats)
# Observation: Business class mean is ~₹52,540 vs Economy mean ~₹6,572 (an 8x difference).
# This is a genuine aviation pricing structure, not an artifact.
```

## 2. Airline and Booking Window Trends


```python
airline_stats = clean_df.groupby('airline')['price'].agg(['count', 'mean', 'median', 'std'])
print(airline_stats.sort_values('mean', ascending=False))

days_left_trend = clean_df.groupby('days_left')['price'].mean()
print('Price by days_left sample (1, 15, 30, 45):')
print(days_left_trend.loc[[1, 15, 30, 45]])
```

## 3. Route Coverage Analysis


```python
routes = clean_df.groupby(['source_city', 'destination_city']).size()
print(f'Unique tier-1 routes: {len(routes)}')
print(routes.head(10))
```
